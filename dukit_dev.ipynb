```python

import pandas as pd
import glob

files_dukit = [f for f in glob.glob('src\\dukit\\**\\*.py', recursive=True)]
files_tests = [f for f in glob.glob('tests\\**\\*.py', recursive=True)]

loc_dukit = 0
loc_tests = 0
locs_dukit = pd.DataFrame(columns=['file', 'lines'])
locs_tests = pd.DataFrame(columns=['file', 'lines'])


for file in files_dukit:
    with open(file, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        loc_dukit += len(lines)
        locs_dukit.loc[len(locs_dukit), :] = [file, len(lines)]
locs_dukit.loc[len(locs_dukit), :] = ['total', loc_dukit]


for file in files_tests:
    with open(file, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        loc_tests += len(lines)
        locs_tests.loc[len(locs_tests), :] = [file, len(lines)]
locs_tests.loc[len(locs_tests), :] = ['total', loc_tests]


locs_dukit = (
    locs_dukit
    .sort_values('lines', ascending=False)
    .reset_index(drop=True)
    .style
    .set_properties(**{'text-align': 'left'})  #type: ignore
    )
locs_tests = (
    locs_tests
    .sort_values('lines', ascending=False)
    .reset_index(drop=True)
    .style
    .set_properties(**{'text-align': 'left'})  #type: ignore
    )


display(locs_dukit)
display(locs_tests)
```
